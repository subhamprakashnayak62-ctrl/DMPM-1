**MIN MAX NORMALIZATION**


```python
import numpy as np
data = [10, 20, 30, 40, 50]
min_value = min(data)
max_value = max(data)
normalized = [''
    (x - min_value) / (max_value - min_value)
    for x in data
]
print("Min-Max normalized:", normalized)
```

**Z-SCORE NORMALIZATION**

```python
import numpy as np
data = [20, 30, 40, 50, 60]
mean = np.mean(data)
std = np.std(data)
normalized = [
    (x - mean) / std
    for x in data
]
print("Mean:", mean)
print("Standard deviation:", std)
print("Z-score normalized:", normalized)
```

**DECIMAL SCALING**

```python
import numpy as np
data = [100, 200, 300, 400, 500]
max_value = max(abs(x) for x in data)
j = len(str(int(max_value)))
normalized = [
    x / (10 ** j)
    for x in data
]
print("Decimal scaling:", normalized)
```

**CORRELATION AND CO VARIANCE**

```python
import numpy as np
X = [10, 20, 30, 40, 50]
Y = [15, 25, 35, 45, 60]
correlation = np.corrcoef(X, Y)[0, 1]
print("Correlation Coefficient:", correlation)
```

```python
import pandas as pd
data = pd.DataFrame({
    'X': [10, 20, 30, 40, 50],
    'Y': [15, 25, 35, 45, 60]
})
print("Covariance:")
print(data.cov())
```

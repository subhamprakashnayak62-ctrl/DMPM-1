**with function**

```python
data = list(map(int, input("Enter the numbers of the list: ").split()))

mean = sum(data) / len(data)

data.sort()
n = len(data)

if n % 2 == 1:
    median = data[n // 2]
else:
    median = (data[n // 2 - 1] + data[n // 2]) / 2
mode = max(set(data), key=data.count)
variance=0

for i in data:
  variance+=(i-mean)**2

variance=variance/(n-1)

std_deviation=variance**0.5

print("Data:", data)
print("Mean:", mean)
print("Median:", median)
print("Mode:", mode)
print("Variance:", variance)
print("Standard Deviation:", std_deviation)
```

**without function**

```python
def mean(lst):
    return sum(lst) / len(lst)

def median(lst):
    lst.sort()
    n = len(lst)

    if n % 2 == 0:
        return (lst[n // 2 - 1] + lst[n // 2]) / 2
    else:
        return lst[n // 2]
def mode(lst):
    max_count = 0
    mode_value = None

    for i in lst:
        count = lst.count(i)
        if count > max_count:
            max_count = count
            mode_value = i

    return mode_value

def variance(lst):
    m = mean(lst)
    return sum((x - m) ** 2 for x in lst) / len(lst)

def standard_deviation(lst):
    return variance(lst) ** 0.5
number = list(map(int, input("Enter numbers: ").split()))

print("Data =", sorted(number))
print("Mean =", mean(number))
print("Median =", median(number))
print("Mode =", mode(number))
print("Variance =", variance(number))
print("Standard Deviation =", standard_deviation(number))
```
